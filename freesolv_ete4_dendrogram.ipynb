Dendrogram of FreeSolv clustering MCES distances with single linkage.

```python
import pickle, sys
from pathlib import Path

repo_root = Path.cwd()
while not (repo_root / "pyproject.toml").exists() and repo_root != repo_root.parent:
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root))

from utils.build_ete4_cluster_cache import cache_path
from utils.ete4_dendrogram_demo import clustering_to_newick, render_mol_images, make_layout
from ete4 import Tree
from ete4.smartview import explorer
from IPython.display import IFrame

DATASET = "freesolv"
with open(cache_path(DATASET), "rb") as fh:
    c = pickle.load(fh)
smiles_list, Z, n_leaves = c["smiles_list"], c["Z"], c["n_leaves"]

image_paths = render_mol_images(
    {str(i): smi for i, smi in enumerate(smiles_list)},
    cache_path(DATASET).parent / f"{DATASET}_full_mol_imgs",
)
tree = Tree(clustering_to_newick(Z[:, :2].astype(int), Z[:, 2], n_leaves, unit_lengths=False), parser=1)
name = f"{DATASET}_full_tree_{n_leaves}mols_reallen"
tree.explore(name=name, layouts=[make_layout(name, image_paths)], keep_server=True, open_browser=False)

host, port = explorer.get_server_address()
IFrame(src=f"http://{host}:{port}/static/gui.html?tree={name}", width=1000, height=800)
```
